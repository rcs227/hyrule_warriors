### Get list of humming .wav files

```python
from remotezip import RemoteZip

url = "https://huggingface.co/datasets/dadinghh2/HumTrans/resolve/main/all_wav.zip"

with RemoteZip(url, timeout=30) as z:
    wav_names = [n for n in z.namelist() if n.lower().endswith(".wav")]
    print("WAV files:", len(wav_names))
    print(*wav_names[:10000], sep="\n")
```

### Print associated MIDI notes

```python
from huggingface_hub import hf_hub_download
from zipfile import ZipFile
from pathlib import Path
import pretty_midi

# Downloads the small MIDI archive, not the large WAV archive
midi_zip = hf_hub_download(
    repo_id="dadinghh2/HumTrans",
    repo_type="dataset",
    filename="all_midi.zip",
)

audio_name = "M04_0061_0001_2_D.wav"  # Replace with your WAV filename
midi_name = Path(audio_name).with_suffix(".mid").name

with ZipFile(midi_zip) as z:
    matches = [n for n in z.namelist() if Path(n).name == midi_name]
    if not matches:
        raise FileNotFoundError(f"No matching MIDI: {midi_name}")

    midi_bytes = z.read(matches[0])

# pretty_midi reads from a file path, so write this small MIDI to a temporary file
temp_mid = Path("one_hum.mid")
temp_mid.write_bytes(midi_bytes)

midi = pretty_midi.PrettyMIDI(str(temp_mid))
for instrument in midi.instruments:
    for note in instrument.notes:
        print(
            pretty_midi.note_number_to_name(note.pitch),
            "MIDI:", note.pitch,
            "onset:", round(note.start, 3),
            "offset:", round(note.end, 3),
        )

temp_mid.unlink()
```

### Download associated MIDI file

```python
from pathlib import Path
from remotezip import RemoteZip

url = "https://huggingface.co/datasets/dadinghh2/HumTrans/resolve/main/all_midi.zip"
audio_name = "M04_0061_0001_2_D.wav"  # Use your exact WAV filename
midi_name = Path(audio_name).with_suffix(".mid").name

with RemoteZip(url, timeout=30) as z:
    matches = [name for name in z.namelist() if Path(name).name == midi_name]
    if not matches:
        raise FileNotFoundError(f"No matching MIDI: {midi_name}")
    Path(midi_name).write_bytes(z.read(matches[0]))

print(f"Saved {midi_name}")
```
